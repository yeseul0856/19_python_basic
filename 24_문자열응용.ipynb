```python
# 문자열 서식 지정자와 포매팅 사용하기
# 변수로써 역할을 할때 %를 적고

name = "james"
print("I am %s" % "james")
print("I am %s" % name)
```

```python
age = 20
print("I am %d years old" % 20)
print("I am %d years old" % age)
```

```python
print("%f" % 2.3)
```

```python
# 서식 지정자로 문자열 정렬하기
# 10칸 확보한 상태에서 오른쪽 정렬
print("%10s" % "python")
# 10칸 확보한 상태에서 왼쪽 정렬
print("%-10s" % "python")
```

```python
# 문자열 안에 값 여러개 넣기
print("Today is %d %s" % (3, "Aprill"))
```

```python
# format 메서드 사용하기
# {0}이 변수다
# 문자열이든 숫자든 format 함수에 인수로 적어두면 된다

print("Hello, {0}".format("world"))
print("Hello, {0}".format(100))
```

```python
# 숫자 순서대로 인수가 들어간다 
print("Hello, {0} {2} {1}".format("Python", "Script", 3.6))
```

```python
print("{0} {0} {1} {1}".format("python", "script"))
```

```python
# 24.6 심사문제: 높은 가격순으로 출력하기
s = list(map(int, input().split(";")))
#51900;83000;158000;367500;250000;59200;128500;1304000

s.sort(reverse=True)

# 10자리를 확보하고 소수점 4개 까지는 :10.4f
for i in s:
    print(f'{i:>9,}')
```
