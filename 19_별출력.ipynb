```python
# 중첩 루프 사용하기

for i in range(5):
    for j in range(5):
        print('j:', j, sep="", end=" ")
    print("i:", i, "\\n", sep="")
```

```python
# 중첩 루프 사용하기

for i in range(5):
    for j in range(5):
        print('j:', j, sep="")  #end=" "가 없다면 
    print("i:", i, "\\n", sep="")
```

```python
# 중첩 for문을 사용하는 예시
fruits = ["사과", "바나나"]
colors = ["빨강", "노랑"]

for fruit in fruits:
    for color in colors:
        print(color, end=" ")
    print(fruit)

    
```

```python
# 중첩 for문을 사용하는 예시
fruits = ["사과", "바나나"]
colors = ["빨강", "노랑"]

for fruit in fruits:
    for color in colors:
        print(color, fruit)
```

```python
# 사각형으로 별 출력하기
for i in range(5):
    for j in range(5):
        print('*', end="") # end=로 인해 옆으로 25개의 별
```

```python
# 사각형으로 별 출력하기
for i in range(5):
    for j in range(5):
        print('*', end="")
    print() #j에서 5개 찍고 print()을 해서 다음줄로 간다
```

```python
# 사각형 모양 바꾸기
for i in range(3):
    for j in range(7):
        print("*", end="")
    print()
```

```python
# 계단식으로 별 출력하기

for i in range(5):
    for j in range(5):
        if j  <= i:
            print("*", end="")
    print()
```

```python
# 대각선으로 별 출력하기

for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# 19.5 연습문제: 역삼각형 모양으로 별 출력하기

for i in range(5):
    for j in range(5):
        if j >= i:
            print('*', end="")
        else:
            print(end=" ")
    print()
```

```python
# 19.6 심사문제: 산 모양으로 별 출력하기

height = int(input())

for i in range(height):
    print(" " * (height-i-1) + "*" * (2*i+1))
```
