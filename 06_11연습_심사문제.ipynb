```python
# 07 출력 방법 알아보기
# 7.4 연습문제: 날짜와 시간 출력하기
year = 2000
month = 10
day = 27
hour = 11
minute = 43
second = 59

print(year, month, day, sep="/", end=' ')
print(hour, minute, second, sep=":")
```

```python
# 7.5 심사문제: 날짜와 시간 출력하기
year, month, day = map(int, input("날짜를 입력하세요").split())
hour, minute, second = map(int, input("시간을 입력하세요").split())

print(year, month, day, sep="-", end='T')
print(hour, minute, second, sep=":")
```
