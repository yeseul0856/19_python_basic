```python
# while 반복문으로 요소 출력하기
a = [38, 31, 53, 62, 19]

i = 0
while i < len(a):
    print(a[i])
    i += 1
```

```python
# 가장 작은 수와 가장 큰 수 구하기
a = [38, 31, 53, 62, 19]

smallest = a[0]

for i in a: 
    if i < smallest:
        smallest = i

print(smallest)
```

```python
# 가장 큰 수 구하기

a = [38, 31, 53, 62, 19]

largest = a[0]

for i in a: 
    if i > largest:
        largest = i

print(largest)
```

```python
# 22.10 심사문제: 2의 거듭제곱 리스트 생성하기
f, s = map(int, input().split())

result = [2 ** i for i in range(f, s+1)]
result.pop(1)
result.pop(-2)
print(result)
```
