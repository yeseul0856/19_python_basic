```python
# 인덱스와 요소를 함께 출력하기

a = [38, 31, 53, 62, 19]

for index, value in enumerate(a):
    print(index, value)
```

```python
# index에 +1을 하면 0부터 시작하지 않고 1부터 시작할수 있다 

a = [38, 31, 53, 62, 19]

for index, value in enumerate(a):
    print(index + 1, value)
```

```python
# while 반복문으로 요소 출력하기
a = [38, 31, 53, 62, 19]

i = 0
while i < len(a):
    print(a[i])
    i += 1
```

```python
# 가장 작은 수와 가장 큰 수 구하기
a = [38, 31, 53, 62, 19]

smallest = a[0]

for i in a: 
    if i < smallest:
        smallest = i

print(smallest)
```

```python
# 가장 큰 수 구하기

a = [38, 31, 53, 62, 19]

largest = a[0]

for i in a: 
    if i > largest:
        largest = i

print(largest)
```

```python
# 22.9 연습문제: 리스트에서 특정 요소만 뽑아내기

a = ['alpha', 'bravo', 'charlie', 'delta', 'echo', 'foxtrot', 'golf', 'hotel', 'india']
b = [i for i in a if len(a) == 5]
print(b)
```

```python
# 22.10 심사문제: 2의 거듭제곱 리스트 생성하기
f, s = map(int, input().split())

result = [2 ** i for i in range(f, s+1)]
result.pop(1)
result.pop(-2)
print(result)
```
