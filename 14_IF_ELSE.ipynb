```python
# if 조건문에 동작 방식 알아보기

if True:
    print("참")
else:
    print("거짓")


if False:
    print("참")
else:
    print("거짓")


if None:
    print("참")
else:
    print("거짓")
```

```python
# 14.6 연습문제: 합격 여부 판단하기
written_test = 75 
coding_test = True

if written_test > 80 and coding_test == True:
    print('합격')
else:
    print("불합격") 
```

```python
# 14.7 심사문제: 합격 여부 판단하기

k, e, m, s = map(int, input().split())
avg = (k + e + m + s) / 4

if 0 <= k <= 100 and 0 <= e <= 100 and 0 <= m <= 100 and 0 <= s <= 100:
    if avg >= 80 :
        print("합격")
    else:
        print("불합격")
else:
    print("잘못된 점수")
```
