```python
# 반복문으로 2차원 리스트의 요소를 출력하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

# i로 출력하면 리스트로 나오는데
for i in a:
    print(i)


# 이런식으로 언패킹을 하면 이렇게 숫자로 나온다
for x, y in a:
    print(x, y)

```

```python
# for 반복문을 두 번 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

# 위에 것과 똑같은 결과 인데 여러 방법으로 사용해보는 중
for i in a:
    for j in i:
        print(j, end=" ")
    print()
```

```python
# for와 range 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j], end=" ")
    print()
```

```python
# while 반복문을 한 번 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

i = 0
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```
