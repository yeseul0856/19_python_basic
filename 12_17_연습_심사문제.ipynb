```python
# 14.7 심사문제: 합격 여부 판단하기
k, e, m, s = map(int, input().split())
avg = (k + e + m + s) / 4

if 0 <= k <= 100 and 0 <= e <= 100 and 0 <= m <= 100 and 0 <= s <= 100:
    if avg >= 80:
        print("합격")
    else: 
        print("불합격")
else: 
    print('잘못된 점수')
```

```python
# 15.3 연습문제: if, elif, else 모두 사용하기
x = int(input())

if 11 <= x <= 20:
    print('11~20')
elif 21 <= x <= 30:
    print('21~30')
else:
    print('아무것도 해당하지 않음')
```

```python
# 15.4 심사문제: 교통카드 시스템 만들기
age = int(input())
balance = 9000

if age >= 7:
    if 7 <= age <= 12:
        balance = balance - 650
    elif 13 <= age <= 18:
        balance = balance - 1050
    elif age >= 19:
        balance = balance - 1250
print(balance)
```

```python
# 17.5 연습문제: 변수 두 개를 다르게 반복하기
i = 2
j = 5

while i <= 32 or j >= 1: 
    print(i, j)
    i *= 2
    j -= 1
```
