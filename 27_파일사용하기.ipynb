```python
# p352
# 파일객체 = open(파일이름, 파일모드)
# 파일객체.write(문자열)
# 파일객체.close()

file = open("hello.txt", "w")
file.write("hello world")
file.close()
```

```python
# 파일에서 문자열 읽기
# csv, 한글, 워드, 엑셀, json 파일 다 가져와서 pandas에서 분석하고 시각화를 할 수 있다

file = open("hello.txt", "r")
s = file.read() # 파일에서 읽은 값을 s에 집어 넣었다
print(s)
file.close()
```

```python
# 자동으로 파일 객체 닫기
# with open("파일 이름", "파일 모드") as 파일 객체: (콜론: 그러면)
#     코드

# 이렇게 적으면 close()를 안적어도 된다

with open("hello2.txt", "w") as file:
    for i in range(3):
        file.write(f"Hello, world! {i}\n") # 이건 print가 아니라 write기 때문에 자동 줄바꿈이 안된다
        # 줄바꿈을 하려면 직접 \n을 적어준다 


```
