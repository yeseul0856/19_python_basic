```python
for i in range(1, 101):
    if i % 3 == 0:
        print("Fizz")
    if i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
for i in range(1, 101):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    if i % 3 == 0:
        print("Fizz")
    if i % 5 == 0:
        print("Buzz")
    else:
        print(i)

# if문에 3의 배수, 5의 배수가 만족할때는(15) FizzBuzz를 출력하는데
# 만약에 두 번째 구문 3의 배수에서 elif가 아니라 if 였다면
# 15에서 FizzBuzz도 나오고 Fizz도(3의 배수) Buzz(5의 배수)도 나오겠지 
# 꼭 elif로 바꾸어 줘야 한다 
```

```python
for i in range(1, 101):
    if i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    elif i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
    else:
        print(i)

# 이렇게 하면 15일때 맨 위에 줄 3의 배수에서 만족하니까 
# Fizz를 출력한다
```

```python
for i in range(1, 101):
    print("Fizz" * (i % 3 == 0) + "Buzz" * (i % 5 == 0) or 1)

# 문자열에 곱하기를 하면 그 숫자만큼 반복이 된다
# 문자열에 false를 곱하게 되면( 문자열 * 0 한것과 같다 )그러면 Fizz도 안나오고
# Buzz도 안나온다 그러면 뒤에 i가 출력된다 
# 3은 (i % 3 == 0)가 true가 되니까 1이 되서 Buzz가 나온다
```

```python
# 20.7 연습문제: 2와 11의 배수, 공배수 처리하기

for i in range(1, 101):
    if i % 2 == 0 and i % 11 == 0:
        print('FizzBuzz')
    elif i % 2 == 0:
        print('Fizz')
    elif i % 11 == 0:
        print('Buzz')
    else:
        print(i)
```

```python
# 20.8 심사문제: 5와 7의 배수, 공배수 처리하기

start, end = map(int, input().split())

if start > end:
    print("잘못된 범위 입니다.")
else:
    for i in range(start, end + 1):
        if i % 35 == 0:
            print('FizzBuzz')
        elif i % 5 == 0:
            print('Fizz')
        elif i % 7 == 0:
            print('Buzz')
        else:
            print(i)
        
```
