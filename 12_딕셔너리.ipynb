```python
# 딕셔너리 만들기
# 딕셔너리 = {
#     키1: 값1,
#     키2: 값2, ...
# }

lux = {
    "health": 490,
    "mana": 334,
    "melee": 550,
    "armor": 18.72
}
lux
```

```python
# 딕셔너리의 키의 자료형
x = { 
    100: "hundred",
    False: 0,
    3.5: [3.5, 3.5]
}
x
```

```python
# 딕셔너리에서 자기자신과 리스트는 키 값으로 사용할 수 없다
x = {
    [10, 20] : 100
}
```

```python
# 튜플은 수정, 추가, 삭제가 안된다
x = {
    (10,20): 100
}
x
```

```python
# 딕셔너리의 키에 접근하고 값 할당하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
print(lux["health"])
print(lux["armor"])
```

```python
# 딕셔너리의 키에 값 할당하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
lux["health"] = 2037
lux["mana"] = 1184
lux
```

```python
lux["mana_regen"] = 3.28
print(lux)
```
