```python
# 위치인수
def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)
```

```python
print_numbers(10, 20, 30)
```

```python
# 언패킹 사용하기
x = [10, 20, 30]
print_numbers(*x)
```

```python
# 가변 인수 함수 만들기

# def 함수이름(*매개변수):
#     코드

def print_numbers(*args):
    print(args)
    for arg in args:
        print(arg)
```

```python
print_numbers(10, 20, 30, 40)
```

```python
# 키워드 인수 

def personal_info(name, age, address):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
personal_info("홍길동", 30, "서울시 용산구")
```

```python
personal_info(age=30, address="서울시", name="홍길동")
```

```python
def personal_info(name, age, address):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
    
```

```python
x = {
    "name": "홍길동",
    "age": 30,
    "address": "서울시 용산구"
}

# personal_info에 3개의 인수를 적어넣어야 하는데 그 세개가 위에 딕셔너리 형태로 x 에 들어있다, 똑같이 3개의 요소니까 나온다
personal_info(**x)
```

```python
# 가변 키워드 인수 함수 만들기
# def 함수이름(**매개변수):
#     코드

# kwargs -> 키워드 아큐먼트 줄임말
def personal_info(**kwargs):
    # print(kwargs)
    for kw, arg in kwargs.items():
        print(kw, arg)
```

```python
personal_info(name="홍길동", age=30, address="서울시 용산구")
```

```python
# 매개변수에 초깃값 지정하기
# "비공개"가 default 값이 되고
# 적으면 적은 값으로 나온다 

def personal_info(name, age, address="비공개"):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
personal_info("홍길동", 30, '서울시 용산구')
```

```python
# 만약에 default값을 넣었다면 그 이후에는 계속 default 값을 넣어 줘야 한다
# 두번째에 default 값을 넣었으니까 세번쨰도 넣어줘야 한다는 오류다 
# 그러니까 default값을 넣을려면 마지막에 적어줘야 한다 

def personal_info(name, address="비공개", age):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```
