```python
# 함수의 결과를 반환하기
# def (함수이름)(매개변수):
#     return 반환값

def add(a, b):
    return a + b
```

```python
x = add(10, 20)

print(x)
```

```python
# 함수에서 값을 여러개 반환하기
def add_sub(a, b):
    return a+b, a-b

x, y = add_sub(10, 20)
print(x, y)
```

```python
# mul a, b는 mul안에서만 사용하는거고
# add a, b는 add안에서만 사용하는건데 
# add, mul에 이름이 같아도 된다(a, b)
def mul(a, b):
    c = a * b
    return c

def add(a, b):
    c = a + b
    print(c)
    d = mul(a, b)
    print(d)

x = 10
y = 20 
add(x, y)

# print(c)가 none이 되는게 아니라
# None은 add(10,20)에 든게 없는거다,
print(add(10,20))
```

```python
# 29.7 연습문제: 몫과 나머지를 구하는 함수 만들기

x = 10
y = 3

def get(a, b):
    return a // b, a % b

quotient, remainder = get(x, y)
print('몫: {0}, 나머지: {1}'.format(quotient,remainder))
```

```python
# 29.8 심사문제: 사칙 연산 함수 만들기 

x, y = map(int, input().split())

def calc(x, y):
    return x + y, x - y, x * y, x / y


a, s, m, d = calc(x,y)
print('덧셈: {0}, 뺄셈: {1}, 곱셈: {2}, 나눗셈: {3}'.format(a, s, m, d))
```
